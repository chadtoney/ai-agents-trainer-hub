# Build a Sequential Workflow with Monitoring in Microsoft Foundry

## Overview
This workshop module teaches participants how to create, configure, and **monitor** a **sequential multi-agent workflow** in Microsoft Foundry portal. You will build a Travel Recommendation System with two specialized agents working together, and learn how to track their behavior in production using Application Insights.

**What You'll Build:**
1. **Front Desk Agent**: Makes attraction recommendations based on user's city interest
2. **Concierge Agent**: Reviews and rates the recommendation with expert analysis
3. **Application Insights Monitoring**: Track agent performance, debug issues, and analyze behavior

## Learning Objectives
By the end of this module, you will be able to:
- Understand sequential workflow patterns and when to use them
- Create multiple agents in Microsoft Foundry portal with specialized instructions
- Configure agents with structured JSON output
- Build a sequential workflow using the Foundry visual workflow builder
- **Set up Application Insights for agent monitoring and observability**
- **Monitor agent behavior, performance, and interactions in the Monitor tab**
- Test and iterate on multi-agent workflows

## Key Benefits of Sequential Orchestration
- **Iterative Refinement**: Second agent improves upon first agent's work
- **Specialization**: Each agent has a specific role and expertise
- **Quality Control**: Built-in review and validation process
- **Clear Information Flow**: Structured handoff between agents

## Why Monitoring Matters
- **Visibility**: Understand how agents behave in production
- **Performance Optimization**: Identify and fix bottlenecks
- **Cost Management**: Track token usage and API costs
- **Debugging**: Quickly diagnose issues with distributed tracing
- **Compliance**: Maintain audit logs of agent interactions

## Prerequisites
- Access to Microsoft Foundry portal
- Azure subscription with permissions to create resources
- Understanding of AI agents (complete Lesson 01 first)

## Exercise 1: Create the Front Desk Agent
**Goal:** Create the first agent that makes attraction recommendations

### Step 1.1: Create Front Desk Agent

**Action Steps:**
1. Navigate to your Foundry project by selecting the dropdown in the top left
2. Ensure your project name is selected
3. Click **"Start building"**, followed by **"Create agent"**
4. Provide agent details:
   - **Agent name**: `frontdeskagent` (use exactly this name)
   - **Description**: `Makes personalized attraction recommendations for travelers`

✅ **Checkpoint:** Front Desk Agent created in the playground

### Step 1.2: Configure Front Desk Agent Instructions

**Action Steps:**
1. In the agent playground, click the **gear icon** in the top right
2. In the left sidebar, find the **"Instructions"** field
3. Copy the instructions below and paste them into the Instructions field

````python
# Front Desk Agent Instructions (Copy and paste into Foundry portal)
"""
You are a friendly Front Desk Agent at a travel hotel. Your role is to make ONE excellent attraction recommendation 
for travelers visiting different cities.

## Your Responsibilities:
1. Ask clarifying questions if the city is not specified
2. Use available tools to gather real-time information:
   - Check current weather conditions
   - Verify attraction opening hours
   - Calculate distances if relevant
3. Make ONE specific attraction recommendation with detailed information
4. Return your recommendation ONLY as valid JSON (no markdown, no code fences)

## Output Format:
You MUST return ONLY valid JSON in this exact format:
{
  "city": "city name",
  "attraction_name": "name of the attraction",
  "description": "detailed description",
  "category": "museum/landmark/park/etc",
  "recommended_duration": "e.g., 2-3 hours",
  "why_recommended": "explanation of why this is recommended",
  "best_time_to_visit": "morning/afternoon/evening with reasoning"
}

## Important Rules:
- ALWAYS use tools when you need real-time information
- Return ONLY the JSON object, no additional text
- No markdown formatting, no code fences (```json)
- Be specific and helpful with your recommendation
"""

print("✅ Copy the instructions above and paste into the Foundry portal Front Desk Agent")
````

### Step 1.3: Test Front Desk Agent Individually

**Action Steps:**
1. In the agent playground chat, type: `"I want to visit an attraction in Barcelona"`
2. Click **"Send"** or press Enter
3. Verify the agent:
   - Provides a specific attraction recommendation
   - Returns valid JSON format
   - Includes all required fields

✅ **Checkpoint:** Front Desk Agent responds with structured JSON recommendation

**If the agent is not working:**
- Ensure the instructions were pasted correctly
- Verify JSON schema was added
- Check that response format is set to "JSON Schema"

## Exercise 2: Create the Concierge Agent
**Goal:** Create the second agent that reviews and rates recommendations

### Step 2.1: Create Concierge Agent

**Action Steps:**
1. Navigate back to your project (click the project name in the top left)
2. Click **"Start building"**, followed by **"Create agent"**
3. Provide agent details:
   - **Agent name**: `conciergeagent` (use exactly this name)
   - **Description**: `Provides expert reviews and ratings for travel attractions`

✅ **Checkpoint:** Concierge Agent created in the playground

### Step 2.2: Configure Concierge Agent Instructions

**Action Steps:**
1. Click the **gear icon** in the top right
2. In the left sidebar, find the **"Instructions"** field
3. Copy the instructions below and paste them into the Instructions field

````python
# Concierge Agent Instructions (Copy and paste into Foundry portal)
"""
You are an expert Concierge Agent with deep knowledge of travel attractions worldwide. Your role is to provide 
professional reviews and ratings for attraction recommendations.

## Your Responsibilities:
1. Receive an attraction recommendation from another agent
2. Use available tools to gather additional information:
   - Get visitor reviews and ratings
   - Compare with alternative attractions
   - Check current ticket prices
3. Provide an expert review with ratings and analysis
4. Return your review ONLY as valid JSON (no markdown, no code fences)

## Output Format:
You MUST return ONLY valid JSON in this exact format:
{
  "attraction_name": "name from the recommendation",
  "city": "city name",
  "popularity_score": 8,
  "popularity_reasoning": "explanation of the score",
  "visitor_rating": 4.5,
  "pros": ["advantage 1", "advantage 2", "advantage 3"],
  "cons": ["drawback 1", "drawback 2"],
  "concierge_recommendation": "your expert opinion and advice",
  "alternative_suggestions": ["alternative 1", "alternative 2"]
}

## Rating Guidelines:
- **popularity_score**: Integer from 1-10 based on overall appeal
- **visitor_rating**: Float from 1.0-5.0 based on visitor feedback
- Provide at least 3 pros and 2 cons
- Suggest 1-3 alternative attractions

## Important Rules:
- ALWAYS use tools to get real visitor data
- Be honest and balanced in your assessment
- Return ONLY the JSON object, no additional text
- No markdown formatting, no code fences (```json)
"""

print("✅ Copy the instructions above and paste into the Foundry portal Concierge Agent")
````

## Exercise 3: Build the Sequential Workflow
**Goal:** Connect both agents in a sequential workflow using the Foundry visual builder

### Step 3.1: Create a New Sequential Workflow

**Action Steps:**
1. Navigate to your project home (click project name in top left)
2. In the left side menu, select **"Workflows"**
3. In the top right of screen, click **"Create"** → **"Sequential"**
4. You should see a sequential workflow template with placeholder agent nodes

**What is a Sequential Workflow?**
- Passes the result from one agent to the next in a defined order
- Perfect for step-by-step processing or multi-stage workflows
- Each agent receives the output of the previous agent

✅ **Checkpoint:** Sequential workflow canvas is open with agent node placeholders

### Step 3.2: Configure the First Agent Node

**Action Steps:**
1. Click on the **first agent node** in the workflow visualizer
2. In the configuration panel that appears:
   - Select **"existing"** (to use an existing agent)
   - Search for **"frontdeskagent"**
   - Select the Front Desk Agent you created earlier
3. Click **"Done"**

✅ **Checkpoint:** First node is now configured with the Front Desk Agent

### Step 3.3: Configure the Second Agent Node

**Action Steps:**
1. Click on the **second agent node** in the workflow visualizer
2. In the configuration panel that appears:
   - Select **"existing"** (to use an existing agent)
   - Search for **"conciergeagent"**
   - Select the Concierge Agent you created earlier
3. Click **"Done"**

✅ **Checkpoint:** Second node is now configured with the Concierge Agent

### Step 3.4: Delete the Third Agent Node 
Note: This is the agent node that was created by default when we created the workflow. 

1. Click the three dots on the agent node. 
2. Select the **"Delete"** button. 

### Step 3.5: Add the End Workflow Action

**Action Steps:**
1. Click the ***"+"*** icon next to the **"conciergeagent"** node. 
2. Scroll down in the pop-up to the **"Basics"** section and choose the **"End"** option. 

### Step 3.5: Save and Name Your Workflow

**Important: Foundry doesn't save workflows automatically!**

**Action Steps:**
1. Click **"Save"** in the workflow visualizer (top toolbar)
2. Give your workflow a meaningful name:
   - **Name**: `attraction-recommendation-workflow`

✅ **Checkpoint:** Workflow is saved with both agents configured in sequential order

2. Complete the **"Exercise: Set Up Agent Monitoring with Application Insights"** section (cells 4-6)
3. Follow the steps to:
   - Navigate to the Monitor tab
   - Create an Application Insights resource
   - Connect it to your Foundry project
4. Return here to continue with Exercise 5

**Quick Setup (if you prefer to do it now):**
1. In the Foundry portal, click the **"Monitor"** tab in the top navigation
2. Click **"Settings"** in the top right
3. Under **"Application insights resource name"** dropdown, select **"Create new resource"**
4. Give it a unique name (e.g., `foundry-workflow-monitoring`)
5. Click **"Create"** and wait 30-60 seconds for provisioning

**Why this matters:**
Application Insights enables you to track workflow performance, debug issues, view token usage, and analyze agent behavior in the Monitor tab. Without it, you won't see telemetry data from your workflow runs.

## Exercise 5: Test the Sequential Workflow

### Step 5.1: Run Your First Workflow

### Step 5.2: Verify Workflow Execution

**What to verify:**
1. **Node completion**: Each node in the visualizer should show as completed (check marks)

   - A specific attraction name
   - Detailed description

   - Alternative suggestions
   - Valid JSON format

**Sequential Flow:**
```
User → Front Desk Agent → Concierge Agent → Final Output
       (Recommendation)    (Expert Review)
```

## Exercise 6: View Monitoring Data in Application Insights
**Goal:** View agent telemetry from your workflow runs in the Monitor tab dashboard

### Step 6.1: Access the Monitor Tab and View Dashboard

**Now that you've run the workflow, let's examine the monitoring data we've been collecting!**

**Remember:** Since we set up Application Insights in Exercise 4, all workflow activity has been automatically tracked.

**Action Steps:**
1. Go back to the **"Monitor"** tab at the top navigation bar
2. Wait 1-2 minutes for telemetry data to appear (initial ingestion delay)
3. You should see an interactive Application Insights dashboard with metrics from your workflow

**Explore the Interactive Dashboard:**
The dashboard displays real-time metrics from your agent workflow. The charts and graphs are fully interactive:

- **Hover over charts**: Move your mouse over any chart or graph to see detailed values at specific points in time
- **View metrics**: See request counts, success rates, and more
- **Adjust time range**: Use the time filter at the top to focus on recent activity (e.g., "Last 30 minutes" or "Last hour")

**What to Look For:**
- Total number of Agent runs (should match the number of times you tested)
- Runs and token metrics
- Tool calls and agent runs
- Error rate

**Try This:**
- Hover over different points on the request chart to see when you ran tests
- Look for the spike in activity when you ran your Barcelona workflow
- Hover over response time charts to see how long each agent took
- Notice any patterns or trends in the data

✅ **Checkpoint:** You can see and interact with monitoring data from your agent workflow
