# Azure AI Agent with Existing Agent Example

This notebook demonstrates working with pre-existing Azure AI Agents by using `provider.get_agent()` method, showing agent reuse patterns for production scenarios.

## Prerequisites
- Azure AI Project created in Azure AI Foundry
- An existing agent named "Travel-agent" (or modify the agent name in the code)
- `AZURE_AI_PROJECT_ENDPOINT` environment variable set
- Azure CLI credentials configured

## Setup: Import Required Libraries

Import all necessary packages for working with Azure AI Agents.

````python
# Copyright (c) Microsoft. All rights reserved.

import asyncio
import os
import time

from agent_framework.azure import AzureAIProjectAgentProvider
from azure.ai.projects.aio import AIProjectClient
from azure.ai.projects.models import PromptAgentDefinition
from azure.identity.aio import AzureCliCredential
from dotenv import load_dotenv
load_dotenv()

# Enable nested asyncio for Jupyter notebooks
import nest_asyncio
nest_asyncio.apply()

print("✅ All imports loaded successfully")
print("✅ Environment variables loaded successfully")
print("✅ Async support enabled for Jupyter")
````

## Define Agent Interaction Function

The following cell **defines** a function that will retrieve an existing Azure AI Agent and ask it multiple questions with delays between them.

**Note:** Running this cell will just define the function (you'll see a green check). To actually execute it and see the results, run the cell below this one that calls `await using_provider_get_agent()`.

````python
async def using_provider_get_agent() -> None:
    """Get an existing Azure AI agent and interact with it."""
    print("=== Get existing Azure AI agent with provider.get_agent() ===\n")

    # Create the client
    async with (
        AzureCliCredential() as credential,
        AIProjectClient(
            endpoint=os.environ["AZURE_AI_PROJECT_ENDPOINT"], 
            credential=credential
        ) as project_client,
    ):
        # Get existing agent by name
        provider = AzureAIProjectAgentProvider(project_client=project_client)
        agent = await provider.get_agent(name="Travel-agent")  # Replace with your actual agent name

        # Verify agent properties
        print(f"Agent ID: {agent.id}")
        print(f"Agent name: {agent.name}")
        print(f"Agent description: {agent.description}\n")

        # List of questions to ask
        questions = [
            "Which destination fits a relaxed food-focused traveler?",
            "What festivals are happening in Rio this month?",
            "I want to visit Paris. What persona best meets this location? Also, what is the closest hotel to the Eiffel Tower?",
        ]

        # Ask each question with a delay
        for i, query in enumerate(questions, 1):
            print(f"\n{'='*60}")
            print(f"Question {i} of {len(questions)}")
            print(f"{'='*60}")
            print(f"User: {query}")
            print()
            
            result = await agent.run(query)
            print(f"Agent: {result}")
            
            # Wait 5 seconds before next question (except after the last one)
            if i < len(questions):
                print("\n⏳ Waiting 5 seconds before next question...")
                time.sleep(5)
````

## Run the Agent Interaction

Execute the function to connect to your Azure AI Agent and ask the questions.

**Note:** Make sure you have:
1. Set the `AZURE_AI_PROJECT_ENDPOINT` environment variable
2. Created an agent named "Travel-agent" in Azure AI Foundry (or update the agent name in the code above)
3. Authenticated with Azure CLI (`az login`)

````python
# Run the agent interaction
await using_provider_get_agent()
````

## Troubleshooting

If you encounter issues:

1. **Environment Variable Error**: Ensure `AZURE_AI_PROJECT_ENDPOINT` is set
   ```python
   import os
   print(os.environ.get("AZURE_AI_PROJECT_ENDPOINT", "Not set"))
   ```

2. **Agent Not Found**: Verify the agent name matches your Azure AI Foundry agent

3. **Authentication Error**: Run `az login` in your terminal to authenticate

4. **Import Errors**: Ensure all packages are at compatible versions (all `1.0.0b260123`)
