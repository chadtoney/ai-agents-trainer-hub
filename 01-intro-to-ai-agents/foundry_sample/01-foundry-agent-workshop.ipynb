## Exercise 2: Add File Search Tool for Knowledge Grounding
**Goal:** Use the File Search tool to upload product data and ground your agent in knowledge

### What is File Search?

The **File Search tool** in Microsoft Foundry augments your agent with knowledge from documents you upload. It functions similarly to Azure AI Search but is built directly into Foundry for simplicity. 

When you upload files, Foundry automatically:
- Chunks the documents
- Creates vector embeddings
- Enables semantic search capabilities
- Allows the agent to retrieve relevant information

This grounds your agent in domain-specific knowledge without requiring external Azure AI Search resources.

### Step 2.1: Upload File Using File Search Tool

**Action Steps:**
1. Navigate to the `Travel-agent` in the Microsoft Foundry Playground
2. In the agent customizable panel, find the **"Tools"** section (left sidebar)
3. In the **"Tools"** section, choose **"Upload files"**
4. In the dropdown, ensure **"Create a new index"** option is selected. We will use the default name provided
5. Click **"browse for files"**
6. Select `data/AI_Travel_Agent_Leisure_Knowledge_Base.txt`
7. Wait for processing to complete:
   - The system chunks the document
   - Creates vector embeddings automatically
   - Makes content searchable for the agent

✅ **Checkpoint:** The file should be uploaded and indexed

### Step 2.2: Verify File Search is Enabled

**Action Steps:**
1. Confirm the File Search tool shows as active
2. Verify your uploaded file appears in the tool's file list
3. The agent can now semantically search this knowledge base to answer queries

✅ **Checkpoint:** File Search tool is active and file is listed

### Step 2.3: Update Agent Instructions

## Exercise 3: Add Bing Grounding Tool
**Goal:** Add web intelligence capabilities using Bing Search

### Step 3.1: Create Bing Grounding Resource

**Action Steps:**
1. Navigate to [Azure Portal](https://portal.azure.com)
2. Search for "Grounding with Bing Search"
3. Create new resource:
   - Select subscription and resource group
   - Choose pricing tier
   - Complete creation
4. Note the resource details for connection

✅ **Checkpoint:** Bing Grounding resource is created in Azure

### Step 3.2: Connect Bing Grounding to Agent

**Action Steps:**
1. Return to Microsoft Foundry portal
2. Open your `Travel-agent`
3. In the agent customizable panel, find the **"Tools"** section (left sidebar)
4. Choose the **"Add a new tool"** option, followed by selecting the **"Grounding with Bing Search"** tool
5. Choose **"Add tool"**
6. In the **"Grounding with Bing Search connection"** field, select your Bing connection
7. Click **"Add"**
8. Save agent

✅ **Checkpoint:** Bing Grounding tool is connected to your agent

### Step 3.3: Update Agent Instructions with Bing Capabilities

**Action:** Copy the comprehensive instructions below and replace the existing instructions

```python
# Final Agent Instructions with Multiple Tools (Copy this text and paste into Foundry portal)
"""
You are an AI Travel Agent.

Scope:
- You may ONLY answer travel-related questions.
- Travel-related includes leisure travel planning, destinations, itineraries, travel styles, pacing, seasonal considerations, and traveler preferences.
- Business, corporate, or non-travel topics are OUT OF SCOPE.

Out-of-Scope Handling:
- If a user question is not travel-related, respond exactly with:
  "My apologies, but I can't assist with this request."

Tool Usage Rules (Mandatory):
1. For leisure travel questions that require reasoning, patterns, personas, or destination knowledge:
   - ALWAYS use the File Search tool as your primary source.
   - Ground your response in retrieved document content.
   - Do not answer from memory alone.

2. For travel questions that require:
   - Current information (events, weather, entry rules, festivals)
   - General travel inspiration not covered in the files
   - Broad destination discovery or comparisons
   - Exploratory travel questions beyond the document
   - Use Bing Grounding tool.

3. If a travel question could be answered by BOTH:
   - Use File Search FIRST.
   - Supplement with Bing Grounding ONLY if the file does not contain sufficient information.

Response Rules:
- Never answer a travel question without using at least one search tool.
- Do not invent facts not present in retrieved sources.
- Do not repeat document text verbatim—generate new itineraries and explanations inspired by it.
- Prefer relaxed, leisure-focused planning language.
"""

print("✅ Copy the instructions above and paste into the Foundry portal")
print("📝 Note: These instructions now include guidance for both File Search and Bing Grounding")
```

**Key Enhancement:** Notice how these instructions build on Exercise 2 by adding external intelligence while keeping internal knowledge via File Search.

## Key Concepts Review

### Agent Tools & Knowledge Grounding

**Knowledge Tools** (provide data/context):
- **File Search**: Augments agent with knowledge from uploaded documents
  - Built-in tool in Azure AI Foundry
  - Creates vector embeddings automatically
  - Provides semantic search capabilities (similar to Azure AI Search)
  - No separate Azure resource needed
- **Grounding with Bing Search**: Provides external, real-time information from the web

**How Tool Selection Works:**
- The agent analyzes user queries and automatically selects appropriate tools
- Instructions guide the model on when to use each tool
- Tools can be invoked multiple times per query for comprehensive answers
- Example: "What tents do we sell?" → File Search | "Competitor tent prices" → Bing Search

### Agent Instructions

Well-crafted instructions should:
- Guide agent behavior and tone
- Define tool usage strategies
- Provide clear scope boundaries
- Set boundaries for appropriate responses

### Progressive Enhancement

The development approach used in this workshop:
1. Start simple with base agent
2. Add tools incrementally
3. Test after each addition
4. Iterate on instructions based on testing

This approach helps you understand the impact of each component and troubleshoot issues effectively.
